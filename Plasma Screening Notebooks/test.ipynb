```python
#IMPORTS
import lammps
import numpy as np
import plasmapy
import matplotlib.pyplot as plt
import pyvista as pv
from astropy import units as u
```

```python
# ── 1. LAMMPS check ──────────────────────────────────────────
lmp = lammps.lammps()
print(f"✓ LAMMPS loaded successfully")
lmp.close()
```

```python
# ── 2. PlasmaPy check ────────────────────────────────────────
from plasmapy.formulary import Debye_length
n_e = 1e19 * u.m**-3       # electron number density
T_e = 1 * u.eV             # electron temperature
debye = Debye_length(T_e, n_e)
print(f"✓ PlasmaPy working — Debye Length: {debye:.4f}")
```

```python
# ── 3. Matplotlib check ───────────────────────────────────────
x = np.linspace(0, 10, 200)
y = np.exp(-x) / x          # rough Yukawa-like decay
plt.figure(figsize=(6, 3))
plt.plot(x, y, color='royalblue')
plt.title("Yukawa Potential (test plot)")
plt.xlabel("r"); plt.ylabel("V(r)")
plt.tight_layout()
plt.show()
print("✓ Matplotlib working")
```

```python
# ── 4. PyVista check ─────────────────────────────────────────
sphere = pv.Sphere()
sphere.plot(color='orange', show_edges=True)
print("✓ PyVista working")

print("\n🎉 All systems go — environment is ready!")
```
